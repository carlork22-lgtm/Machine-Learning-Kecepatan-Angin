```python
y = df["wind_speed_10m_max (km/h)"]

print("Target (y):")
print(y.head())
```

```python
print(df.columns.tolist())
```

```python
X = df[
    [
        "weather_code (wmo code)",
        "temperature_2m_mean (°C)",
        "temperature_2m_max (°C)",
        "temperature_2m_min (°C)",
        "wind_direction_10m_dominant (°)",
        "surface_pressure_mean (hPa)",
        "relative_humidity_2m_mean (%)",
        "cloud_cover_mean (%)",
        "shortwave_radiation_sum (MJ/m²)"
    ]
]

print("Shape X:", X.shape)
print("Shape y:", y.shape)
```

```python
print("X:")
display(X.head())

print("\ny:")
display(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    shuffle=False
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Periode data training:")
print(X_train.index.min(), "sampai", X_train.index.max())

print("\nPeriode data testing:")
print(X_test.index.min(), "sampai", X_test.index.max())
```

```python
print("Tanggal training:")
print(df.loc[X_train.index, "time"].min(), "sampai", df.loc[X_train.index, "time"].max())

print("\nTanggal testing:")
print(df.loc[X_test.index, "time"].min(), "sampai", df.loc[X_test.index, "time"].max())
```

DATA SPLITTING

```python
from sklearn.model_selection import train_test_split

# Split data 70% training dan 30% testing
X_train_70, X_test_70, y_train_70, y_test_70 = train_test_split(
    X,
    y,
    test_size=0.3,
    shuffle=False
)

print("Pembagian Data 70:30")
print("X_train :", X_train_70.shape)
print("X_test  :", X_test_70.shape)
print("y_train :", y_train_70.shape)
print("y_test  :", y_test_70.shape)

print("\nTanggal Training:")
print(
    df.loc[X_train_70.index, "time"].min(),
    "sampai",
    df.loc[X_train_70.index, "time"].max()
)

print("\nTanggal Testing:")
print(
    df.loc[X_test_70.index, "time"].min(),
    "sampai",
    df.loc[X_test_70.index, "time"].max()
)
```

MEMBUAT MODEL RANDOM FOREST

```python
from sklearn.ensemble import RandomForestRegressor

rf_70 = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_70.fit(X_train_70, y_train_70)

print("Random Forest 70:30 selesai dilatih.")
```

```python
# Melakukan prediksi pada data testing
y_pred_rf_70 = rf_70.predict(X_test_70)

# Membuat tabel aktual vs prediksi
hasil_rf_70 = pd.DataFrame({
    "Tanggal": df.loc[X_test_70.index, "time"],
    "Aktual (km/h)": y_test_70.values,
    "Prediksi (km/h)": y_pred_rf_70
})

print("Jumlah data prediksi:", len(y_pred_rf_70))

display(hasil_rf_70.head(10))
```

```python
import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae_rf_70 = mean_absolute_error(y_test_70, y_pred_rf_70)
rmse_rf_70 = np.sqrt(mean_squared_error(y_test_70, y_pred_rf_70))
r2_rf_70 = r2_score(y_test_70, y_pred_rf_70)

print("Evaluasi Random Forest 70:30")
print(f"MAE  : {mae_rf_70:.4f} km/h")
print(f"RMSE : {rmse_rf_70:.4f} km/h")
print(f"R²   : {r2_rf_70:.4f}")
```

```python
plt.figure(figsize=(15, 5))

plt.plot(
    df.loc[X_test_70.index, "time"],
    y_test_70.values,
    label="Aktual"
)

plt.plot(
    df.loc[X_test_70.index, "time"],
    y_pred_rf_70,
    label="Prediksi Random Forest"
)

plt.title("Random Forest 70:30 - Aktual vs Prediksi")
plt.xlabel("Tanggal")
plt.ylabel("Kecepatan Angin (km/h)")
plt.legend()
plt.grid(True)

plt.show()
```

SPLIT DATA 70:30 XGBoost
